## Understanding of what the data represents
### Drug response
- ln_ic50 represents the log of concentration where drug kills 50% of cells. lower is better.
- auc represents overall drug response. high auc means that the cells may not die after a point of high dosage also

### RNASeq data
Basically there are some issues with raw data:
- Some genes are long so there is a chance you may find shorter genes more
- some cell lines are sequenced more deeply (more reads - more chunks) which leads to higher probability of finding some genes
#### what is in the files
- merged -> for every cell line x gene the expected raw count, the tpm, the fpkm 
- expected count -> the raw counts of the times a gene was seen in the cell line
- fpkm -> fragments per kilobase per million -> essentially normalizes for gene length and sequencing depth. but for every CLxGene pair we can't do 1to1 comparison because genes have different lengths and depths are different
- tpm -> transcripts per million -> divides by gene length and then makes sure that for every column (cell line) we can do 1to1 comparisons, makes sure all gene normalized counts are normalized so the cell lne count sums to 1 million

## Generate relevant data

```python
drugs_chosen =  ['Erlotinib', 'Gefitinib', 'Tamoxifen', 'Paclitaxel', 'Gemcitabine']
```

```python
tpm_transpose = rsem_tpm.transpose()
```

```python
tpm_transpose.columns = tpm_transpose.iloc[0]
tpm_transpose=tpm_transpose[3::]
```

```python
fitted.head()
```

```python
tpm_transpose.index.name = 'model_id'
tpm_transpose = tpm_transpose.reset_index()
```

```python
tpm_transpose.head()
```

```python
filtered_drugs = fitted[fitted['DRUG_NAME'].isin(drugs_chosen)]
```

```python
drugs_response_tpm_merged_data = pd.merge(
    left = tpm_transpose,
    right=filtered_drugs,
    left_on='model_id',
    right_on="SANGER_MODEL_ID")
```
